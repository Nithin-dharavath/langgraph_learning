```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict, Literal
```

```python
class StudentInput(TypedDict):
    name : str
    marks : int
    message : str
```

```python
def pass_result(state : StudentInput):
    return{
        "message" : "you have passed the Exam"
        }
```

```python
def fail_result(state : StudentInput):
    return{
        "message" : "you have fail the Exam! Better luck next time"
        }
```

```python
def check_result(state : StudentInput) -> Literal["pass_result", "fail_result"]:
    marks = state["marks"]
    if marks >= 40:
        return "pass_result"
    else: 
        return "fail_result"
```

```python
def route_check(state : StudentInput):
    marks = state["marks"]
    if marks >= 40:
        return {"message" : "your are eligible for future"}
    return {"message" : "your are not eligible for future"}
```

```python
graph = StateGraph(StudentInput)

graph.add_node("check_result", check_result)
graph.add_node("route_check", route_check)
graph.add_node("pass_result", pass_result)
graph.add_node("fail_result", fail_result)

graph.add_edge(START, "route_check")
graph.add_conditional_edges("route_check", check_result)
graph.add_edge("pass_result", END)
graph.add_edge("fail_result", END)

workflow = graph.compile()
```

```python
initial_state = {
    "name" : "Nithin",
    "marks" : 41,
}
final_state = workflow.invoke(initial_state)
print(final_state)
```
